# Data Cleaning: COE Bidding Results and Vehicle Population

This notebook profiles the two LTA datasets in `../data/` and checks them for quality problems before analysis:

1. `COEBiddingResultsPrices.csv`: the result of each COE bidding exercise, 2010–2026
2. `AnnualMotorVehiclePopulationbyVehicleType.csv`: yearly vehicle population by type, 2005–2024

For each dataset:
- check raw file's structure and text formatting
- profile the numeric columns
- check the unit, the time coverage and perform reasonableness check
- flag suspicious values

Post-review, apply the necessary data transformation, and save the cleaned data as new files in `data/` (with the suffix `_clean`). The original files are not changed.

````python
# imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

pd.set_option('display.width', 200)
pd.set_option('display.max_columns', 30)
pd.set_option('display.max_rows', 100)

DATA_DIR = '../data'
COE_PATH = f'{DATA_DIR}/COEBiddingResultsPrices.csv'
POP_PATH = f'{DATA_DIR}/AnnualMotorVehiclePopulationbyVehicleType.csv'
COE_CLEAN_PATH = f'{DATA_DIR}/COEBiddingResultsPrices_clean.csv'
POP_CLEAN_PATH = f'{DATA_DIR}/AnnualMotorVehiclePopulationbyVehicleType_clean.csv'
````

````python
# Chart style: light background, quiet grid and axes, one blue for the data
INK, INK_2, MUTED, GRID, AXIS, SURFACE = '#0b0b0b', '#52514e', '#898781', '#e1e0d9', '#c3c2b7', '#fcfcfb'
BLUE, BLUE_LIGHT = '#2a78d6', '#b7d3f6'

plt.rcParams.update({
    'figure.facecolor': SURFACE, 'axes.facecolor': SURFACE, 'savefig.facecolor': SURFACE,
    'axes.edgecolor': AXIS, 'axes.labelcolor': INK_2, 'axes.titlecolor': INK,
    'axes.titlesize': 10, 'axes.titleweight': 'bold', 'axes.titlelocation': 'left',
    'axes.spines.top': False, 'axes.spines.right': False,
    'axes.grid': True, 'axes.grid.axis': 'y', 'axes.axisbelow': True,
    'grid.color': GRID, 'grid.linewidth': 0.6,
    'xtick.color': AXIS, 'ytick.color': AXIS, 'xtick.labelcolor': INK_2, 'ytick.labelcolor': INK_2,
    'font.size': 9, 'figure.dpi': 110, 'figure.titlesize': 12, 'figure.titleweight': 'bold',
})
THOUSANDS = FuncFormatter(lambda x, _: f'{x:,.0f}')


def hist_panel(ax, s, title, bins=40, log=False):
    """Draw a histogram of `s` on chart panel `ax`, with a dashed line and label at the median.
    Used to see the shape of a numeric field (skew, peaks, outliers); missing values are dropped.
    Set log=True when values span several orders of magnitude, such as vehicle population."""
    s = s.dropna()
    if log:
        bins = np.logspace(np.log10(s.min()), np.log10(s.max()), bins)
        ax.set_xscale('log')
    ax.hist(s, bins=bins, color=BLUE, edgecolor=SURFACE, linewidth=0.8)
    med = s.median()
    ax.axvline(med, color=INK_2, linewidth=1, linestyle='--')
    ax.annotate(f'median {med:,.0f}', xy=(med, 1), xycoords=('data', 'axes fraction'),
                xytext=(4, -2), textcoords='offset points', va='top', fontsize=8, color=INK_2)
    ax.set_title(title)
    ax.xaxis.set_major_formatter(THOUSANDS)
    ax.set_ylabel('rows')


def box_by_group(ax, df, value, group, title):
    """Draw one box plot of column `value` for each group in column `group` on chart panel `ax`.
    Used to compare a metric across COE categories: the box is the middle 50%, the dark line
    is the median, and grey dots are outliers beyond 1.5 × the box height."""
    groups = sorted(df[group].unique())
    data = [df.loc[df[group] == g, value].dropna() for g in groups]
    ax.boxplot(data, patch_artist=True, widths=0.55,
               boxprops=dict(facecolor=BLUE_LIGHT, edgecolor=BLUE, linewidth=1),
               medianprops=dict(color=INK, linewidth=1.5),
               whiskerprops=dict(color=BLUE, linewidth=1), capprops=dict(color=BLUE, linewidth=1),
               flierprops=dict(marker='o', markersize=3, markerfacecolor=MUTED, markeredgecolor='none', alpha=0.6))
    ax.set_xticks(range(1, len(groups) + 1), [g.replace('Category ', '') for g in groups])
    ax.set_xlabel('Category')
    ax.set_title(title)
    ax.yaxis.set_major_formatter(THOUSANDS)
````

### 1.1 Raw text checks
Ingest every column as text to view formatting problems (blanks, whitespace, thousands separators).

````python
# read raw coe bidding prices file as text
coe_raw = pd.read_csv(COE_PATH, dtype=str)
print(coe_raw.shape)
coe_raw.head()
````

````python
def text_profile(df, int_cols=()):
    """Count formatting problems in each column of a raw DataFrame read with dtype=str.
    Reports nulls, blanks, distinct values, stray spaces and commas, plus values in `int_cols`
    that are still not whole numbers after removing commas. Returns one row per column."""
    rows = []
    for col in df.columns:
        s = df[col]
        row = {
            'column': col,
            'nulls': s.isna().sum(),
            'empty': (s.str.strip() == '').sum(),
            'distinct': s.nunique(),
            'leading/trailing ws': (s != s.str.strip()).sum(),
            'has comma': s.str.contains(',', na=False).sum(),
        }
        if col in int_cols:
            row['not an integer (after removing commas)'] = (~s.str.replace(',', '').str.fullmatch(r'-?\d+').fillna(False)).sum()
        rows.append(row)
    return pd.DataFrame(rows).set_index('column')

text_profile(coe_raw, int_cols=['bidding_no', 'quota', 'bids_success', 'bids_received', 'premium'])
````

- There are no missing values/nulls, empty strings or stray whitespaces. 
- There are thousands separators in `bids_success` (41 rows) and `bids_received` (109 rows). The **separators need to be removed, and the column be typed as integer after**.

### 1.6 Suspicious values
We don't expect huge jumps between one exercise and the next in the same category so if there is, it likely points to a data quality issue.

````python
# Find percentage change in premium from one exercise to the next, for each COE category. 
coe = coe.sort_values(['vehicle_class', 'month', 'bidding_no']).reset_index(drop=True)
coe['premium_pct_chg'] = coe.groupby('vehicle_class')['premium'].pct_change()

print('Exercise-to-exercise % change in premium:')
print(coe['premium_pct_chg'].describe(percentiles=[.05, .5, .95]).round(3).to_string())
coe.loc[coe['premium_pct_chg'].abs().nlargest(6).index, COLS + ['premium_pct_chg']]
````

- Typically we'd expect about 0.5% change in premiums from one exercise to the next across all categories
- Only for Jan 2010 does Cat D have a significantly high premium of 20K (a 22% change from the previous exercise). As Cat typically are priced lower, this could potentially be a data quality issue and needs to be investigated before the analysis.

````python
# Investigating the outlier data found earlier by looking at the first two months, all categories side by side
coe[coe['month'].isin(['2010-01', '2010-02'])].sort_values(['month', 'bidding_no', 'vehicle_class'])[COLS]
````

**These 2 values identified earlier look like they were erroneously copied from the wrong row:**

| Row | Suspicious value | Copied from | Details of Potential Error |
|---|---|---|---|
| 2010-01, exercise 2, **Category D** | `premium` = 20,090 | Category C premium in the same exercise (20,090) | Category D was 889 the exercise before and 852 the one after. That shows as +2,160% then −96%. |
| 2010-02, exercise 1, **Category B** | `quota` = 1,154 | Category A quota in the same exercise (1,154) | Category B's quota is about 690 in the exercises around it. The wrong value creates the only undersubscribed row and the largest unallocated gap. |

Both should be checked against LTA's published results. Until then, it is wise to leave them as NA so they don't affect the statistics. It can be imputed reasonably (for example, with an average of the two values before and after the exercise), but imputing it upfront into a cleaned file would treat that number as a fact as much as the rest of the numbers when it isn't. In other words, no one would know that this value is imputed. Therefore, leave it as NA in the cleaned file, but for any kind of analysis that involved the `premium` or `quota` ,it can be imputed then.

### 1.8 Relationships between categories and columns

Cat D is only about 0.5 correlated wtih everything. This strongly points to how the motorcycle market has it's own demand.

````python
# Correlation with premium within each category (Spearman), so category size doesn't dominate
pd.DataFrame({
    k: g[['quota', 'bids_success', 'bids_received', 'bid_ratio']].corrwith(g['premium'], method='spearman')
    for k, g in coe.groupby('vehicle_class')
}).T.round(2)
````

- Across all rows, `quota` vs `premium` correlation is about zero. **Within each category** it is clearly negative, strongest for E (−0.83) and B (−0.50) which means **a smaller quota means a higher premium**. Category D is the exception (+0.29).
- `bid_ratio` is only weakly related to premium within each category (0.05 to 0.22). This means that knowing how oversubscribed an exercise is doesn't really tell us anything about the premium.

### 1.9 Yearly view and first vs second exercise

````python
# Revenue is the total amount paid for COEs in each exercise (premium x number of successful bids). We can look at revenue by year, and also the average premium by year for each category.
coe['revenue'] = coe['premium'] * coe['bids_success']
yr = coe['date'].dt.year

yearly = coe.groupby(yr).agg(quota=('quota', 'sum'), bids_success=('bids_success', 'sum'),
                             bids_received=('bids_received', 'sum'), revenue_bn=('revenue', 'sum'))
yearly['revenue_bn'] /= 1e9
for cat in ['A', 'B']:
    sub = coe[coe['vehicle_class'] == f'Category {cat}']
    yearly[f'avg_premium_{cat}'] = sub.groupby(sub['date'].dt.year)['premium'].mean()
yearly.round(2)
````

````python
# Yearly trends as small multiples, one scale per panel.
# 2020 is low because bidding was suspended April to June. The last year is partial, so totals are shown lighter and dashed.
ORANGE = '#eb6834'
last = yearly.index[-1]
full = yearly.index[:-1]
fig, axes = plt.subplots(1, 3, figsize=(14, 3.8))

ax = axes[0]
ax.bar(yearly.index, yearly['revenue_bn'], width=0.8, edgecolor=SURFACE, linewidth=1,
       color=[BLUE] * len(full) + [BLUE_LIGHT])
ax.annotate('partial year', xy=(last, yearly['revenue_bn'].iloc[-1]), xytext=(0, -4),
            textcoords='offset points', ha='center', va='top', rotation=90, fontsize=8, color=INK_2)
ax.set_title('COE revenue (SGD billion)')

# Totals: solid for full years, dashed into the partial year
ax = axes[1]
for col, label, color, dy in [('quota', 'quota', BLUE, 0), ('bids_received', 'bids received', ORANGE, 0)]:
    ax.plot(full, yearly.loc[full, col], color=color, linewidth=2, label=label)
    ax.plot(yearly.index[-2:], yearly[col].iloc[-2:], color=color, linewidth=2, linestyle='--')
    ax.annotate(label, xy=(last, yearly[col].iloc[-1]), xytext=(4, dy),
                textcoords='offset points', va='center', fontsize=8, color=INK_2)
ax.set_title('Supply vs demand (COEs per year)')

# Averages for a partial year
ax = axes[2]
for col, label, color, dy in [('avg_premium_A', 'Category A', BLUE, -6), ('avg_premium_B', 'Category B', ORANGE, 6)]:
    ax.plot(yearly.index, yearly[col], color=color, linewidth=2, label=label)
    ax.annotate(label, xy=(last, yearly[col].iloc[-1]), xytext=(4, dy),
                textcoords='offset points', va='center', fontsize=8, color=INK_2)
ax.set_title('Average premium (SGD)')

for ax in axes:
    ax.set_xticks(range(yearly.index[0], last + 1, 2))
    ax.axvline(2020, color=MUTED, linewidth=0.8, linestyle=':')
for ax in axes[1:]:
    ax.legend(frameon=False, loc='upper left', fontsize=8)
    ax.yaxis.set_major_formatter(THOUSANDS)
    ax.set_xlim(right=last + 3)
axes[0].annotate('2020: no bidding\nApr to Jun', xy=(2020, 1), xycoords=('data', 'axes fraction'),
                 xytext=(-4, -2), textcoords='offset points', ha='right', va='top', fontsize=8, color=INK_2)
fig.suptitle('COE bidding: yearly trends', x=0.01, ha='left')
fig.tight_layout()
plt.show()
````

- Supporting the negative correlation between `quota` and `premium`, the line charts show that both of them move in opposite directions.
- Supply or `quota` peaked at 2016-2017, while `premium` had their lowest point in 2018-2019.
- Premiums have increased significantly since the dip, while quota maintained fairly low. Revenue reached a high of ~$6BN in 2025, contributed by the high premiums.

````python
# Median % change in premium from one exercise to the next in the same month, by category
q = coe.pivot_table(index=['vehicle_class', 'month'], columns='bidding_no', values='premium')
((q[2] - q[1]) / q[1]).groupby(level=0).median().rename('median % change, exercise 1 → 2').map('{:.2%}'.format)
````

- The two exercises in a month differ by less than 1% at the median in every category. Taking into consideration that 2020 has only 9 months and 2026 only 9 months (to September).

````python
# One small chart per vehicle type, grouped by category, over the years. Each panel has its own y-axis starting at 0, so the lines show how each type changed relative to its own size.
import textwrap
series = pop.assign(label=np.where(pop['category'] == 'Tax Exempted Vehicles', 'Tax exempted: ' + pop['type'], pop['type']))
groups = list(series.groupby(['category', 'label'], sort=True))
ncols = 3
nrows = -(-len(groups) // ncols)
fig, axes = plt.subplots(nrows, ncols, figsize=(11, 2.7 * nrows), sharex=True, layout='constrained')
for ax, ((cat, label), g) in zip(axes.flat, groups):
    ax.plot(g['year'], g['number'], color=BLUE, linewidth=2)
    ax.set_ylim(bottom=0, top=g['number'].max() * 1.15)
    ax.set_title(textwrap.fill(label, 34), fontsize=10)
    ax.annotate(cat, xy=(0, 0), xycoords='axes fraction', xytext=(4, 4), textcoords='offset points', fontsize=8, color=MUTED)
    ax.yaxis.set_major_formatter(FuncFormatter(lambda x, _: f'{x/1000:g}k' if x >= 1000 else f'{x:,.0f}'))
    ax.yaxis.set_major_locator(plt.MaxNLocator(3))
    ax.set_xticks([2005, 2015, 2024])
    ax.tick_params(labelsize=8.5, labelbottom=True)
for ax in axes.flat[len(groups):]:
    ax.set_visible(False)
fig.suptitle('Vehicle population by type over time (each panel on its own scale, from 0)', x=0.01, ha='left')
fig.get_layout_engine().set(h_pad=0.12, w_pad=0.15)
plt.show()
````

- *Private cars* stayed around 500,000 from about 2010, which fits the COE cap on car numbers. It peaked in 2013.
- From 2013, Rental Cars are split into *Private Hire (Chauffeur) cars* and *Private Hire (Self-Drive) cars*. Both of these two categories grew to about 61K and 36K respectively, while *Taxis* fell by more than half (~15K) after 2014.
- *Off peak cars* peaks around 2009 and then falls steadily. *Private hire buses* also fell sharply.

Note: Private Hire (Self-Drive) refers to a rental car where the person renting it drives it. Private Hire (Chauffeur) on the other hand, is a car used to carry paying passengers eg. Grab, Gojek, etc. For this, drivers needs a Private Hire Car Driver's Vocational Licence (PDVL)

In 2012, taxi companies bidding in Cat A were seen as pushing up prices. They were then taken out of bidding and now pay the Prevailing Quota Premium (PQP) instead. However, Private Hire Cars were not treated the same (although the reality is that more taxi drivers have switched to private hire chaffeuring).

Since there's no separate COE category for private hire cars, fleet operators eg. Grab/Gojek etc are businesses which are bidding against households. These companies can pass on the COE cost through rental fees but not household car users. As private hire growth was a large share of car COE demand in 2015 to 2017 and in 2013 to 2024, economists have linked this to higher permiums. While this stance is argued by critics, it is not proven. Business bidders win only 6-8% of car COEs which turned out to be small % of the population of bidders.

### 2.4 Vehicle Population by category

````python
# Pivot table of vehicle population by category
by_cat = pop.pivot_table(index='year', columns='category', values='number', aggfunc='sum')
by_cat['TOTAL'] = by_cat.sum(axis=1)
by_cat.style.format('{:,.0f}')
````

````python
# Growth of each category since the first year (first year = 100), all panels on the same scale.
# The grey line is total vehicles, for comparison.
idx = by_cat / by_cat.iloc[0] * 100
first, last = idx.index[0], idx.index[-1]
cats = [c for c in idx.columns if c != 'TOTAL']
ncols = 3
nrows = -(-len(cats) // ncols)
fig, axes = plt.subplots(nrows, ncols, figsize=(11, 3 * nrows), sharex=True, sharey=True, layout='constrained')
for ax, cat in zip(axes.flat, cats):
    ax.axhline(100, color=AXIS, linewidth=0.8)
    ax.plot(idx.index, idx['TOTAL'], color=MUTED, linewidth=1.2, label='All vehicles')
    ax.plot(idx.index, idx[cat], color=BLUE, linewidth=2, label=cat)
    change = idx[cat].iloc[-1] - 100
    ax.annotate(f'{change:+.0f}%', xy=(last, idx[cat].iloc[-1]), xytext=(4, 0),
                textcoords='offset points', va='center', fontsize=9, color=INK_2)
    ax.set_title(f'{cat}\n{by_cat[cat].iloc[0]:,.0f} → {by_cat[cat].iloc[-1]:,.0f}', fontsize=10)
    ax.set_xticks([first, 2010, 2015, 2020, last])
    ax.set_xlim(right=last + 3)
    ax.tick_params(labelsize=8.5, labelbottom=True)
for ax in axes.flat[len(cats):]:
    ax.set_visible(False)
axes.flat[0].legend(handles=axes.flat[0].get_lines()[1:2], frameon=False, loc='upper left', fontsize=8.5)
fig.suptitle(f'Vehicle population by category: growth since {first} ({first} = 100)', x=0.01, ha='left')
plt.show()
````

- The total number of vehicles grew fast to about 970,000 in 2012, then stayed almost flat until 2020 as the allowed vehicle growth rate was cut (to zero for cars and motorcycles from 2018). It passed 1 million in 2024.
- *Taxis* fell by 41% from 2005, and by more than half from their 2014 peak, as ride-hailing grew.
- *Cars* grew 50%. Most of that was before 2013, with a smaller rise after 2016 that includes private hire cars.
- *Tax Exempted Vehicles* almost doubled (+90%), the fastest growth of any category.
- *Motorcycles* (+6%) and *goods vehicles* (+12%) barely changed.

- The biggest jumps are real market changes: private-hire chauffeur cars growing after ride-hailing arrived, and off-peak cars shrinking. 
- The 2013 falls in **GPVs (−23.6%)** and **tax-exempted motorcycles (−23.9%)** happen in the same year the car types were redefined. They are likely reclassifications, so treat 2012→2013 as a break in the series.

````python
# Pivot table of vehicle population by type, for Cars and Station-wagons only
cars = pop[pop['category'] == 'Cars and Station-wagons'].pivot_table(index='year', columns='type', values='number')
cars
````

- Private Hire (Chaffeur) grew from 1.6K in 2014 to 46.9K in 2017, an increase of 45K in 3 years.
- In the same period, private cars fell from 536K to 502K (ie. ~35K drop)
- Both of this occuring within the same time period suggests that some drivers moved from owning a car to driving for or riding with ride-hail services.

````python
# Change in vehicle numbers since 2013 (when Private Hire types start), on one axis in the same unit.
# If car owners moved to ride-hailing, private cars and taxis should fall while chauffeured private hire cars rise.
BASE = 2013
ORANGE, AQUA = '#eb6834', '#1baf7a'
shift = pd.DataFrame({
    'Private Hire (Chauffeur) cars': cars['Private Hire (Chauffeur) cars'],
    'Private cars': cars['Private cars'],
    'Taxis': pop[pop['type'] == 'Taxis'].set_index('year')['number'],
}).loc[BASE:]
change = shift - shift.loc[BASE]

fig, ax = plt.subplots(figsize=(11, 4.2))
ax.axvspan(2014, 2017, color=GRID, alpha=0.5, linewidth=0)
ax.annotate('ride-hailing boom\n2014 to 2017', xy=(2015.5, 1), xycoords=('data', 'axes fraction'),
            xytext=(0, -4), textcoords='offset points', ha='center', va='top', fontsize=8.5, color=INK_2)
ax.axhline(0, color=AXIS, linewidth=0.8)
for col, color, dy in zip(change.columns, [ORANGE, BLUE, AQUA], [0, -7, 7]):  # dy keeps close end labels apart
    ax.plot(change.index, change[col], color=color, linewidth=2, marker='o', markersize=4, label=col)
    end = change[col].iloc[-1]
    ax.annotate(f'{col}  {end:+,.0f}', xy=(change.index[-1], end), xytext=(6, dy),
                textcoords='offset points', va='center', fontsize=8.5, color=INK_2)
ax.yaxis.set_major_formatter(FuncFormatter(lambda x, _: f'{x:+,.0f}' if x else '0'))
ax.set_xticks(range(BASE, change.index[-1] + 1))
ax.set_xlim(BASE - 0.3, change.index[-1] + 4.5)
ax.set_ylabel(f'change in vehicles since {BASE}')
ax.legend(frameon=False, loc='lower left', fontsize=8.5)
ax.set_title(f'Private cars and taxis fell as chauffeured private hire cars grew (change since {BASE})')
fig.tight_layout()
plt.show()
````

- From **2014 to 2017**, chauffeured private hire cars rose by about 46K while private cars fell by about 38K. The two changes are similar in size and happen in the same years. Private cars fell even though the COE quota was high in 2015 to 2017, so fewer people chose to replace their cars.
- **Taxis** start falling from 2016 and are down about 14.6K by 2024.
- **After 2017** private cars partly recover, but private hire stays high. By 2024, private hire is up about 59K, while private cars and taxis together are down about 30K.
- As these are net population counts, they can't show whether the same people switched. Some private hire cars are converted private cars, and others are new cars. The chart supports a shift toward ride-hailing but doesn't prove causation

## Part 4: Apply fixes and save
The two suspicious values are set to missing (`pd.NA`). The cleaned data is saved as new `_clean.csv` files in `data/`. The original files are not changed.

````python
coe_clean = pd.read_csv(COE_PATH, thousands=',')
coe_clean['date'] = pd.to_datetime(coe_clean['month'])
coe_clean[NUM] = coe_clean[NUM].astype('Int64')   # nullable integers, so missing values are allowed

fixes = [
    # (month, bidding_no, vehicle_class, column)
    ('2010-01', 2, 'Category D', 'premium'),
    ('2010-02', 1, 'Category B', 'quota'),
]
for month, bid, cls, col in fixes:
    mask = (coe_clean['month'] == month) & (coe_clean['bidding_no'] == bid) & (coe_clean['vehicle_class'] == cls)
    assert mask.sum() == 1
    print(f'{month} #{bid} {cls}: {col} {coe_clean.loc[mask, col].item()} -> <NA>')
    coe_clean.loc[mask, col] = pd.NA

coe_clean[NUM].isna().sum()
````
